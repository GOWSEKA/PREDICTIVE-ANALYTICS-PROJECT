```python
# ============================================================
# EFFICIENTNET-B4 - MELANOMA CLASSIFICATION
#
# TRAINED COMPLETELY FROM SCRATCH
# NO PRETRAINED WEIGHTS
# NO TRANSFER LEARNING
# NO EARLY STOPPING
#
# DATASET:
#   5229 valid lesions
#   450 Melanoma
#   4779 Non-Melanoma
#
# IMAGES:
#   10458 total
#   2 images per lesion
#
# SPLIT:
#   Training   : 3660 lesions = 7320 images
#   Validation :  784 lesions = 1568 images
#   Testing    :  785 lesions = 1570 images
#
# MODEL:
#   EfficientNet-B4
#
# EPOCHS:
#   50
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import copy
import random

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import Dataset, DataLoader

from torchvision import transforms
from torchvision.models import efficientnet_b4

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    roc_auc_score,
    auc
)

from tqdm.auto import tqdm


# ============================================================
# 2. REPRODUCIBILITY
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

print("Random seed:", SEED)


# ============================================================
# 3. DEVICE
# ============================================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("\n========================================")
print("DEVICE")
print("========================================")

print("Using device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("GPU not available. Using CPU.")


# ============================================================
# 4. PATHS
# ============================================================

BASE_PATH = (
    "/content/drive/MyDrive/"
    "MELANOMA/PREDICTIVE PROJECT"
)

# IMPORTANT:
# This is your NEW full-dataset preparation folder.
PREPARED_DATA_PATH = os.path.join(
    BASE_PATH,
    "implement",
    "prepared_data_B4"
)

OUTPUT_PATH = os.path.join(
    BASE_PATH,
    "implement",
    "B4_results"
)

os.makedirs(
    OUTPUT_PATH,
    exist_ok=True
)


# ============================================================
# 5. CSV FILES
# ============================================================

TRAIN_CSV = os.path.join(
    PREPARED_DATA_PATH,
    "train_images.csv"
)

VAL_CSV = os.path.join(
    PREPARED_DATA_PATH,
    "validation_images.csv"
)

TEST_CSV = os.path.join(
    PREPARED_DATA_PATH,
    "test_images.csv"
)


# ============================================================
# 6. CHECK FILES
# ============================================================

print("\n========================================")
print("CHECKING DATA FILES")
print("========================================")

print("Train CSV:", TRAIN_CSV)
print("Validation CSV:", VAL_CSV)
print("Test CSV:", TEST_CSV)

if not os.path.exists(TRAIN_CSV):
    raise FileNotFoundError(
        "Training CSV not found!"
    )

if not os.path.exists(VAL_CSV):
    raise FileNotFoundError(
        "Validation CSV not found!"
    )

if not os.path.exists(TEST_CSV):
    raise FileNotFoundError(
        "Testing CSV not found!"
    )

print("All CSV files found.")


# ============================================================
# 7. LOAD CSV FILES
# ============================================================

train_df = pd.read_csv(TRAIN_CSV)
val_df = pd.read_csv(VAL_CSV)
test_df = pd.read_csv(TEST_CSV)

print("\n========================================")
print("DATASET INFORMATION")
print("========================================")

print("Training images:", len(train_df))
print("Validation images:", len(val_df))
print("Testing images:", len(test_df))


# ============================================================
# 8. DISPLAY COLUMNS
# ============================================================

print("\nTrain columns:")
print(train_df.columns.tolist())

print("\nFirst training rows:")
print(train_df.head())


# ============================================================
# 9. DATASET CLASS
# ============================================================

class MelanomaDataset(Dataset):

    def __init__(
        self,
        dataframe,
        transform=None
    ):

        self.dataframe = (
            dataframe.reset_index(drop=True)
        )

        self.transform = transform

    def __len__(self):

        return len(self.dataframe)

    def __getitem__(self, index):

        row = self.dataframe.iloc[index]

        image_path = row["image_path"]

        # Your prepared CSV uses binary_label
        label = int(row["binary_label"])

        image = Image.open(
            image_path
        ).convert("RGB")

        if self.transform:
            image = self.transform(image)

        return (
            image,
            label,
            row["lesion_id"],
            image_path
        )


# ============================================================
# 10. IMAGE TRANSFORMS
# ============================================================

IMAGE_SIZE = 224


# Training normalization calculated from your
# full training dataset during preparation.

MEAN = [
    0.6786853261290028,
    0.5268318828827355,
    0.4761676111750107
]

STD = [
    0.1258003819658543,
    0.1335618838464332,
    0.15567726312182673
]


train_transform = transforms.Compose([

    transforms.Resize(256),

    transforms.RandomResizedCrop(
        IMAGE_SIZE,
        scale=(0.85, 1.0),
        ratio=(0.95, 1.05)
    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomVerticalFlip(
        p=0.2
    ),

    transforms.RandomRotation(
        15
    ),

    transforms.ColorJitter(
        brightness=0.1,
        contrast=0.1,
        saturation=0.1,
        hue=0.02
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=MEAN,
        std=STD
    )
])


val_test_transform = transforms.Compose([

    transforms.Resize(
        (IMAGE_SIZE, IMAGE_SIZE)
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=MEAN,
        std=STD
    )
])


# ============================================================
# 11. CREATE DATASETS
# ============================================================

train_dataset = MelanomaDataset(
    train_df,
    transform=train_transform
)

val_dataset = MelanomaDataset(
    val_df,
    transform=val_test_transform
)

test_dataset = MelanomaDataset(
    test_df,
    transform=val_test_transform
)


# ============================================================
# 12. DATA LOADERS
# ============================================================

BATCH_SIZE = 16
NUM_WORKERS = 0


train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available()
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available()
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available()
)


# ============================================================
# 13. CLASS DISTRIBUTION
# ============================================================

print("\n========================================")
print("CLASS DISTRIBUTION")
print("========================================")

print("\nTraining:")
print(
    train_df["binary_label"]
    .value_counts()
    .sort_index()
)

print("\nValidation:")
print(
    val_df["binary_label"]
    .value_counts()
    .sort_index()
)

print("\nTesting:")
print(
    test_df["binary_label"]
    .value_counts()
    .sort_index()
)


# ============================================================
# 14. EFFICIENTNET-B4 PARAMETERS
# ============================================================

NUM_CLASSES = 2
DROPOUT = 0.4


print("\n========================================")
print("EFFICIENTNET-B4 PARAMETERS")
print("========================================")

print("Image size       :", IMAGE_SIZE)
print("Number classes   :", NUM_CLASSES)
print("Dropout          :", DROPOUT)
print("Pretrained       : NONE")
print("Transfer learning: NO")


# ============================================================
# 15. CREATE EFFICIENTNET-B4
# ============================================================

# IMPORTANT:
# weights=None means the model starts from random weights.
# NO ImageNet pretrained weights are used.

model = efficientnet_b4(
    weights=None
)


# Replace original classifier

in_features = model.classifier[1].in_features

model.classifier = nn.Sequential(

    nn.Dropout(
        p=DROPOUT
    ),

    nn.Linear(
        in_features,
        NUM_CLASSES
    )
)


# ============================================================
# 16. MOVE MODEL TO DEVICE
# ============================================================

model = model.to(device)


# ============================================================
# 17. MODEL INFORMATION
# ============================================================

total_parameters = sum(
    p.numel()
    for p in model.parameters()
)

trainable_parameters = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)


print("\n========================================")
print("MODEL INFORMATION")
print("========================================")

print(
    "Model              : EfficientNet-B4"
)

print(
    "Total parameters   :",
    total_parameters
)

print(
    "Trainable parameters:",
    trainable_parameters
)

print(
    "Pretrained weights : NONE"
)

print(
    "Training method    : FROM SCRATCH"
)


# ============================================================
# 18. CLASS WEIGHTS
# ============================================================

class_counts = (
    train_df["binary_label"]
    .value_counts()
    .sort_index()
)

num_class_0 = class_counts.get(
    0,
    0
)

num_class_1 = class_counts.get(
    1,
    0
)

total_samples = (
    num_class_0 +
    num_class_1
)


weight_0 = total_samples / (
    2 * num_class_0
)

weight_1 = total_samples / (
    2 * num_class_1
)


class_weights = torch.tensor(
    [
        weight_0,
        weight_1
    ],
    dtype=torch.float32
).to(device)


print("\n========================================")
print("CLASS WEIGHTS")
print("========================================")

print(
    "Non-Melanoma weight:",
    weight_0
)

print(
    "Melanoma weight    :",
    weight_1
)


# ============================================================
# 19. FOCAL LOSS
# ============================================================

class WeightedFocalLoss(nn.Module):

    def __init__(
        self,
        alpha=None,
        gamma=2.0
    ):

        super().__init__()

        self.alpha = alpha
        self.gamma = gamma

    def forward(
        self,
        inputs,
        targets
    ):

        ce_loss = F.cross_entropy(
            inputs,
            targets,
            weight=self.alpha,
            reduction="none"
        )

        pt = torch.exp(
            -ce_loss
        )

        focal_loss = (
            (1 - pt) ** self.gamma
        ) * ce_loss

        return focal_loss.mean()


criterion = WeightedFocalLoss(
    alpha=class_weights,
    gamma=2.0
)


# ============================================================
# 20. OPTIMIZER
# ============================================================

LEARNING_RATE = 0.0001
WEIGHT_DECAY = 0.0001


optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)


# ============================================================
# 21. LEARNING RATE SCHEDULER
# ============================================================

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="max",
    factor=0.5,
    patience=3
)


# ============================================================
# 22. TRAINING SETTINGS
# ============================================================

EPOCHS = 50


print("\n========================================")
print("TRAINING SETTINGS")
print("========================================")

print("Epochs       :", EPOCHS)
print("Batch size   :", BATCH_SIZE)
print("Learning rate:", LEARNING_RATE)
print("Weight decay :", WEIGHT_DECAY)

print("\nNO EARLY STOPPING")


# ============================================================
# 23. TRAINING HISTORY
# ============================================================

history = {

    "train_loss": [],
    "train_accuracy": [],
    "train_precision": [],
    "train_recall": [],
    "train_f1": [],

    "val_loss": [],
    "val_accuracy": [],
    "val_precision": [],
    "val_recall": [],
    "val_f1": [],
    "val_roc_auc": [],

    "learning_rate": []
}


# ============================================================
# 24. BEST MODEL VARIABLES
# ============================================================

best_val_roc_auc = -1

best_model_state = copy.deepcopy(
    model.state_dict()
)


# ============================================================
# 25. TRAINING FUNCTION
# ============================================================

def train_one_epoch(
    model,
    loader,
    criterion,
    optimizer,
    device
):

    model.train()

    running_loss = 0.0

    all_predictions = []
    all_labels = []

    for images, labels, _, _ in tqdm(
        loader,
        desc="Training",
        leave=False
    ):

        images = images.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )

        optimizer.zero_grad()

        outputs = model(images)

        loss = criterion(
            outputs,
            labels
        )

        loss.backward()

        optimizer.step()

        running_loss += (
            loss.item() *
            images.size(0)
        )

        predictions = torch.argmax(
            outputs,
            dim=1
        )

        all_predictions.extend(
            predictions.detach()
            .cpu()
            .numpy()
        )

        all_labels.extend(
            labels.detach()
            .cpu()
            .numpy()
        )


    epoch_loss = (
        running_loss /
        len(loader.dataset)
    )


    epoch_accuracy = accuracy_score(
        all_labels,
        all_predictions
    )

    epoch_precision = precision_score(
        all_labels,
        all_predictions,
        zero_division=0
    )

    epoch_recall = recall_score(
        all_labels,
        all_predictions,
        zero_division=0
    )

    epoch_f1 = f1_score(
        all_labels,
        all_predictions,
        zero_division=0
    )


    return (
        epoch_loss,
        epoch_accuracy,
        epoch_precision,
        epoch_recall,
        epoch_f1
    )


# ============================================================
# ============================================================
# 26. VALIDATION FUNCTION
# ============================================================

def validate_one_epoch(
    model,
    loader,
    criterion,
    device
):

    model.eval()

    running_loss = 0.0

    all_predictions = []
    all_labels = []
    all_probabilities = []

    with torch.no_grad():

        for images, labels, _, _ in tqdm(
            loader,
            desc="Validation",
            leave=False
        ):

            images = images.to(
                device,
                non_blocking=True
            )

            labels = labels.to(
                device,
                non_blocking=True
            )

            outputs = model(images)

            loss = criterion(
                outputs,
                labels
            )

            running_loss += (
                loss.item() *
                images.size(0)
            )

            probabilities = torch.softmax(
                outputs,
                dim=1
            )[:, 1]

            predictions = torch.argmax(
                outputs,
                dim=1
            )

            all_predictions.extend(
                predictions.cpu().numpy()
            )

            all_labels.extend(
                labels.cpu().numpy()
            )

            all_probabilities.extend(
                probabilities.cpu().numpy()
            )

    epoch_loss = (
        running_loss /
        len(loader.dataset)
    )

    epoch_accuracy = accuracy_score(
        all_labels,
        all_predictions
    )

    epoch_precision = precision_score(
        all_labels,
        all_predictions,
        zero_division=0
    )

    epoch_recall = recall_score(
        all_labels,
        all_predictions,
        zero_division=0
    )

    epoch_f1 = f1_score(
        all_labels,
        all_predictions,
        zero_division=0
    )

    epoch_roc_auc = roc_auc_score(
        all_labels,
        all_probabilities
    )

    return (
        epoch_loss,
        epoch_accuracy,
        epoch_precision,
        epoch_recall,
        epoch_f1,
        epoch_roc_auc
    )


# 27. TRAINING LOOP
# ============================================================

print("\n\n")
print("========================================")
print("STARTING TRAINING")
print("========================================")


for epoch in range(EPOCHS):

    current_epoch = epoch + 1


    # ----------------------------------------
    # TRAIN
    # ----------------------------------------

    (
        train_loss,
        train_accuracy,
        train_precision,
        train_recall,
        train_f1
    ) = train_one_epoch(
        model,
        train_loader,
        criterion,
        optimizer,
        device
    )


    # ----------------------------------------
    # VALIDATION
    # ----------------------------------------

    (
        val_loss,
        val_accuracy,
        val_precision,
        val_recall,
        val_f1,
        val_roc_auc
    ) = validate_one_epoch(
        model,
        val_loader,
        criterion,
        device
    )


    # ----------------------------------------
    # LEARNING RATE
    # ----------------------------------------

    scheduler.step(
        val_roc_auc
    )

    current_lr = (
        optimizer.param_groups[0]["lr"]
    )


    # ----------------------------------------
    # SAVE HISTORY
    # ----------------------------------------

    history["train_loss"].append(
        train_loss
    )

    history["train_accuracy"].append(
        train_accuracy
    )

    history["train_precision"].append(
        train_precision
    )

    history["train_recall"].append(
        train_recall
    )

    history["train_f1"].append(
        train_f1
    )

    history["val_loss"].append(
        val_loss
    )

    history["val_accuracy"].append(
        val_accuracy
    )

    history["val_precision"].append(
        val_precision
    )

    history["val_recall"].append(
        val_recall
    )

    history["val_f1"].append(
        val_f1
    )

    history["val_roc_auc"].append(
        val_roc_auc
    )

    history["learning_rate"].append(
        current_lr
    )


    # ----------------------------------------
    # SAVE BEST MODEL
    # ----------------------------------------

    if val_roc_auc > best_val_roc_auc:

        best_val_roc_auc = val_roc_auc

        best_model_state = copy.deepcopy(
            model.state_dict()
        )

        torch.save(
            best_model_state,
            os.path.join(
                OUTPUT_PATH,
                "best_efficientnet_b4_model.pth"
            )
        )

        best_text = " <-- BEST"

    else:

        best_text = ""


    # ----------------------------------------
    # PRINT RESULTS
    # ----------------------------------------

    print(
        f"Epoch [{current_epoch:02d}/{EPOCHS}] | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_accuracy:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val ROC-AUC: {val_roc_auc:.4f} | "
        f"LR: {current_lr:.7f}"
        f"{best_text}"
    )


# ============================================================
# 28. LOAD BEST MODEL
# ============================================================

print("\n========================================")
print("LOADING BEST MODEL")
print("========================================")

model.load_state_dict(
    best_model_state
)

print(
    "Best Validation ROC-AUC:",
    f"{best_val_roc_auc:.4f}"
)


# ============================================================
# 29. SAVE TRAINING HISTORY
# ============================================================

history_df = pd.DataFrame(
    history
)

history_df.to_csv(
    os.path.join(
        OUTPUT_PATH,
        "training_history.csv"
    ),
    index=False
)


# ============================================================
# 30. PLOT TRAINING LOSS
# ============================================================

plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history["train_loss"],
    label="Training Loss"
)

plt.plot(
    history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.title(
    "EfficientNet-B4 Training and Validation Loss"
)

plt.legend()
plt.grid(True)

plt.savefig(
    os.path.join(
        OUTPUT_PATH,
        "loss_curve.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 31. PLOT ACCURACY
# ============================================================

plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history["train_accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.title(
    "EfficientNet-B4 Training and Validation Accuracy"
)

plt.legend()
plt.grid(True)

plt.savefig(
    os.path.join(
        OUTPUT_PATH,
        "accuracy_curve.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 32. PLOT F1 SCORE
# ============================================================

plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history["train_f1"],
    label="Training F1"
)

plt.plot(
    history["val_f1"],
    label="Validation F1"
)

plt.xlabel("Epoch")
plt.ylabel("F1 Score")

plt.title(
    "EfficientNet-B4 Training and Validation F1 Score"
)

plt.legend()
plt.grid(True)

plt.savefig(
    os.path.join(
        OUTPUT_PATH,
        "f1_curve.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 33. TEST EVALUATION
# ============================================================

print("\n========================================")
print("TESTING")
print("========================================")


model.eval()

test_predictions = []
test_probabilities = []
test_labels = []

test_lesion_ids = []
test_image_paths = []


with torch.no_grad():

    for (
        images,
        labels,
        lesion_ids,
        image_paths
    ) in tqdm(
        test_loader,
        desc="Testing"
    ):

        images = images.to(
            device,
            non_blocking=True
        )

        outputs = model(images)

        probabilities = torch.softmax(
            outputs,
            dim=1
        )

        predictions = torch.argmax(
            outputs,
            dim=1
        )


        test_predictions.extend(
            predictions
            .cpu()
            .numpy()
        )

        test_probabilities.extend(
            probabilities[:, 1]
            .cpu()
            .numpy()
        )

        test_labels.extend(
            labels.numpy()
        )

        test_lesion_ids.extend(
            list(lesion_ids)
        )

        test_image_paths.extend(
            list(image_paths)
        )


# ============================================================
# 34. IMAGE-LEVEL METRICS
# ============================================================

image_accuracy = accuracy_score(
    test_labels,
    test_predictions
)

image_precision = precision_score(
    test_labels,
    test_predictions,
    zero_division=0
)

image_recall = recall_score(
    test_labels,
    test_predictions,
    zero_division=0
)

image_f1 = f1_score(
    test_labels,
    test_predictions,
    zero_division=0
)


print("\n========================================")
print("IMAGE-LEVEL RESULTS")
print("========================================")

print(
    "Accuracy :",
    f"{image_accuracy:.4f}"
)

print(
    "Precision:",
    f"{image_precision:.4f}"
)

print(
    "Recall   :",
    f"{image_recall:.4f}"
)

print(
    "F1 Score :",
    f"{image_f1:.4f}"
)


# ============================================================
# 35. CLASSIFICATION REPORT
# ============================================================

print("\n========================================")
print("CLASSIFICATION REPORT")
print("========================================")

print(
    classification_report(
        test_labels,
        test_predictions,
        target_names=[
            "Non-Melanoma",
            "Melanoma"
        ],
        zero_division=0
    )
)


# ============================================================
# 36. IMAGE-LEVEL CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    test_labels,
    test_predictions
)


print("\n========================================")
print("IMAGE-LEVEL CONFUSION MATRIX")
print("========================================")

print(cm)


plt.figure(
    figsize=(6, 5)
)

plt.imshow(cm)

plt.title(
    "EfficientNet-B4 Image-Level Confusion Matrix"
)

plt.xlabel(
    "Predicted Label"
)

plt.ylabel(
    "True Label"
)

plt.xticks(
    [0, 1],
    [
        "Non-Melanoma",
        "Melanoma"
    ]
)

plt.yticks(
    [0, 1],
    [
        "Non-Melanoma",
        "Melanoma"
    ]
)


for i in range(2):

    for j in range(2):

        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center"
        )


plt.colorbar()

plt.savefig(
    os.path.join(
        OUTPUT_PATH,
        "image_confusion_matrix.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 37. IMAGE-LEVEL ROC-AUC
# ============================================================

fpr, tpr, thresholds = roc_curve(
    test_labels,
    test_probabilities
)

roc_auc = auc(
    fpr,
    tpr
)


print("\n========================================")
print("IMAGE-LEVEL ROC-AUC")
print("========================================")

print(
    "ROC-AUC:",
    f"{roc_auc:.4f}"
)


# ============================================================
# 38. ROC CURVE
# ============================================================

plt.figure(
    figsize=(7, 6)
)

plt.plot(
    fpr,
    tpr,
    label=(
        f"EfficientNet-B4 "
        f"ROC-AUC = {roc_auc:.4f}"
    )
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "EfficientNet-B4 ROC Curve"
)

plt.legend()
plt.grid(True)

plt.savefig(
    os.path.join(
        OUTPUT_PATH,
        "roc_curve.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 39. IMAGE-LEVEL PREDICTIONS CSV
# ============================================================

image_predictions_df = pd.DataFrame({

    "lesion_id":
        test_lesion_ids,

    "image_path":
        test_image_paths,

    "true_label":
        test_labels,

    "predicted_label":
        test_predictions,

    "melanoma_probability":
        test_probabilities
})


image_predictions_df.to_csv(
    os.path.join(
        OUTPUT_PATH,
        "image_level_predictions.csv"
    ),
    index=False
)


# ============================================================
# 40. LESION-LEVEL EVALUATION
# ============================================================

print("\n========================================")
print("LESION-LEVEL EVALUATION")
print("========================================")


prediction_df = (
    image_predictions_df.copy()
)


lesion_results = []


for lesion_id, group in prediction_df.groupby(
    "lesion_id"
):

    # Each lesion has two images:
    # clinical + dermoscopic.
    #
    # Average their melanoma probabilities.

    melanoma_probability = (
        group[
            "melanoma_probability"
        ].mean()
    )


    predicted_label = int(
        melanoma_probability >= 0.5
    )


    true_label = int(
        group["true_label"].iloc[0]
    )


    lesion_results.append({

        "lesion_id":
            lesion_id,

        "true_label":
            true_label,

        "melanoma_probability":
            melanoma_probability,

        "predicted_label":
            predicted_label
    })


lesion_df = pd.DataFrame(
    lesion_results
)


# ============================================================
# 41. LESION-LEVEL METRICS
# ============================================================

lesion_accuracy = accuracy_score(
    lesion_df["true_label"],
    lesion_df["predicted_label"]
)

lesion_precision = precision_score(
    lesion_df["true_label"],
    lesion_df["predicted_label"],
    zero_division=0
)

lesion_recall = recall_score(
    lesion_df["true_label"],
    lesion_df["predicted_label"],
    zero_division=0
)

lesion_f1 = f1_score(
    lesion_df["true_label"],
    lesion_df["predicted_label"],
    zero_division=0
)


print(
    "Number of test lesions:",
    len(lesion_df)
)

print(
    "Accuracy :",
    f"{lesion_accuracy:.4f}"
)

print(
    "Precision:",
    f"{lesion_precision:.4f}"
)

print(
    "Recall   :",
    f"{lesion_recall:.4f}"
)

print(
    "F1 Score :",
    f"{lesion_f1:.4f}"
)


# ============================================================
# 42. LESION-LEVEL ROC-AUC
# ============================================================

lesion_auc = auc(
    *roc_curve(
        lesion_df["true_label"],
        lesion_df["melanoma_probability"]
    )[:2]
)


print(
    "ROC-AUC :",
    f"{lesion_auc:.4f}"
)


# ============================================================
# 43. LESION-LEVEL CONFUSION MATRIX
# ============================================================

lesion_cm = confusion_matrix(
    lesion_df["true_label"],
    lesion_df["predicted_label"]
)


print(
    "\nLesion-level confusion matrix:"
)

print(
    lesion_cm
)


plt.figure(
    figsize=(6, 5)
)

plt.imshow(
    lesion_cm
)

plt.title(
    "EfficientNet-B4 Lesion-Level Confusion Matrix"
)

plt.xlabel(
    "Predicted Label"
)

plt.ylabel(
    "True Label"
)

plt.xticks(
    [0, 1],
    [
        "Non-Melanoma",
        "Melanoma"
    ]
)

plt.yticks(
    [0, 1],
    [
        "Non-Melanoma",
        "Melanoma"
    ]
)


for i in range(2):

    for j in range(2):

        plt.text(
            j,
            i,
            lesion_cm[i, j],
            ha="center",
            va="center"
        )


plt.colorbar()

plt.savefig(
    os.path.join(
        OUTPUT_PATH,
        "lesion_confusion_matrix.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 44. LESION-LEVEL ROC CURVE
# ============================================================

lesion_fpr, lesion_tpr, lesion_thresholds = (
    roc_curve(
        lesion_df["true_label"],
        lesion_df["melanoma_probability"]
    )
)


plt.figure(
    figsize=(7, 6)
)

plt.plot(
    lesion_fpr,
    lesion_tpr,
    label=(
        f"EfficientNet-B4 "
        f"Lesion ROC-AUC = "
        f"{lesion_auc:.4f}"
    )
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "EfficientNet-B4 Lesion-Level ROC Curve"
)

plt.legend()
plt.grid(True)

plt.savefig(
    os.path.join(
        OUTPUT_PATH,
        "lesion_roc_curve.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 45. SAVE LESION RESULTS
# ============================================================

lesion_df.to_csv(
    os.path.join(
        OUTPUT_PATH,
        "lesion_level_predictions.csv"
    ),
    index=False
)


# ============================================================
# 46. FINAL RESULTS SUMMARY
# ============================================================

results_summary = pd.DataFrame({

    "Metric": [

        "Image Accuracy",
        "Image Precision",
        "Image Recall",
        "Image F1",
        "Image ROC-AUC",

        "Lesion Accuracy",
        "Lesion Precision",
        "Lesion Recall",
        "Lesion F1",
        "Lesion ROC-AUC"
    ],

    "Score": [

        image_accuracy,
        image_precision,
        image_recall,
        image_f1,
        roc_auc,

        lesion_accuracy,
        lesion_precision,
        lesion_recall,
        lesion_f1,
        lesion_auc
    ]
})


results_summary.to_csv(
    os.path.join(
        OUTPUT_PATH,
        "final_results.csv"
    ),
    index=False
)


# ============================================================
# 47. FINAL OUTPUT
# ============================================================

print("\n")
print("========================================")
print("FINAL RESULTS")
print("========================================")

print(
    results_summary.to_string(
        index=False
    )
)


print("\n========================================")
print("FILES SAVED")
print("========================================")

print(
    "Output folder:",
    OUTPUT_PATH
)

print("\nSaved files:")

for filename in sorted(
    os.listdir(OUTPUT_PATH)
):

    print(
        " -",
        filename
    )


# ============================================================
# 48. COMPLETED
# ============================================================

print("\n========================================")
print("TRAINING AND EVALUATION COMPLETED")
print("========================================")

print(
    "Model: EfficientNet-B4"
)

print(
    "Epochs trained:",
    EPOCHS
)

print(
    "Training lesions: 3660"
)

print(
    "Validation lesions: 784"
)

print(
    "Testing lesions: 785"
)

print(
    "Best Validation ROC-AUC:",
    f"{best_val_roc_auc:.4f}"
)

print(
    "Test Image F1:",
    f"{image_f1:.4f}"
)

print(
    "Test Lesion F1:",
    f"{lesion_f1:.4f}"
)

print(
    "Test Image ROC-AUC:",
    f"{roc_auc:.4f}"
)

print(
    "Test Lesion ROC-AUC:",
    f"{lesion_auc:.4f}"
)
```

```python
from google.colab import drive
drive.mount('/content/drive')

import os
import numpy as np, pandas as pd, torch, torch.nn as nn
from PIL import Image
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from torchvision.models import efficientnet_b4
from sklearn.metrics import (confusion_matrix, roc_auc_score, average_precision_score,
                             brier_score_loss, matthews_corrcoef, cohen_kappa_score,
                             balanced_accuracy_score)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# ---------- paths (same as your training notebook) ----------
BASE_PATH = "/content/drive/MyDrive/MELANOMA/PREDICTIVE PROJECT"
DATA_PATH = os.path.join(BASE_PATH, "implement", "prepared_data_B4")
OUT_PATH  = os.path.join(BASE_PATH, "implement", "B4_results")

val_df  = pd.read_csv(os.path.join(DATA_PATH, "validation_images.csv"))
test_df = pd.read_csv(os.path.join(DATA_PATH, "test_images.csv"))

# ---------- same normalization and transform as training ----------
MEAN = [0.6786853261290028, 0.5268318828827355, 0.4761676111750107]
STD  = [0.1258003819658543, 0.1335618838464332, 0.15567726312182673]

val_test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=MEAN, std=STD)
])

class MelanomaDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.transform = transform
    def __len__(self):
        return len(self.dataframe)
    def __getitem__(self, index):
        row = self.dataframe.iloc[index]
        image = Image.open(row["image_path"]).convert("RGB")
        if self.transform:
            image = self.transform(image)
        return image, int(row["binary_label"]), row["lesion_id"], row["image_path"]

val_loader  = DataLoader(MelanomaDataset(val_df,  val_test_transform), batch_size=16, shuffle=False, num_workers=0)
test_loader = DataLoader(MelanomaDataset(test_df, val_test_transform), batch_size=16, shuffle=False, num_workers=0)

# ---------- rebuild EfficientNet-B4 and load the best saved weights ----------
model = efficientnet_b4(weights=None)
in_features = model.classifier[1].in_features
model.classifier = nn.Sequential(nn.Dropout(p=0.4), nn.Linear(in_features, 2))
model.load_state_dict(torch.load(os.path.join(OUT_PATH, "best_efficientnet_b4_model.pth"),
                                 map_location=device))
model = model.to(device)
model.eval()
print("Best model loaded.")

# ---------- helper functions ----------
from tqdm.auto import tqdm

val_loader  = DataLoader(MelanomaDataset(val_df,  val_test_transform), batch_size=16, shuffle=False, num_workers=2)
test_loader = DataLoader(MelanomaDataset(test_df, val_test_transform), batch_size=16, shuffle=False, num_workers=2)

def lesion_probs(loader):
    model.eval()
    rows = []
    with torch.no_grad():
        for images, labels, lesion_ids, _ in tqdm(loader):
            p = torch.softmax(model(images.to(device)), dim=1)[:, 1].cpu().numpy()
            rows += list(zip(lesion_ids, labels.numpy(), p))
    df = pd.DataFrame(rows, columns=["lesion_id", "label", "prob"])
    return df.groupby("lesion_id").agg(label=("label", "first"),
                                       prob=("prob", "mean")).reset_index()

def metrics(df, t):
    y, p = df["label"].values, df["prob"].values
    pred = (p >= t).astype(int)
    tn, fp, fn, tp = confusion_matrix(y, pred).ravel()
    prec = tp / (tp + fp) if tp + fp else 0
    rec = tp / (tp + fn)
    return dict(threshold=round(t, 2), accuracy=(tp + tn) / len(y), precision=prec, recall=rec,
                specificity=tn / (tn + fp),
                f1=2 * prec * rec / (prec + rec) if prec + rec else 0,
                tn=tn, fp=fp, fn=fn, tp=tp)

# ---------- lesion-level probabilities ----------
val_les  = lesion_probs(val_loader)
test_les = lesion_probs(test_loader)
print("Validation lesions:", len(val_les), "| Test lesions:", len(test_les))

# sanity check: this must reproduce your earlier result (TN 466, FP 251, FN 4, TP 64)
print("Check at 0.50:", metrics(test_les, 0.50))

# ---------- threshold sweep on VALIDATION lesions ----------
sweep = pd.DataFrame([metrics(val_les, t) for t in np.arange(0.30, 0.96, 0.05)])
print(sweep.round(4).to_string(index=False))
best_t = float(sweep.loc[sweep["f1"].idxmax(), "threshold"])
print("\nBest validation threshold:", best_t)

# ---------- final TEST results at the chosen threshold ----------
m = metrics(test_les, best_t)
y, p = test_les["label"].values, test_les["prob"].values
pred = (p >= best_t).astype(int)
print("\nFINAL TEST RESULTS (threshold =", best_t, ")")
for k in ["accuracy", "precision", "recall", "specificity", "f1"]:
    print(f"{k:12s}: {m[k]*100:.2f}%")
print("ROC-AUC     :", round(roc_auc_score(y, p), 4))
print("PR-AUC      :", round(average_precision_score(y, p), 4))
print("Balanced acc:", round(balanced_accuracy_score(y, pred) * 100, 2), "%")
print("MCC         :", round(matthews_corrcoef(y, pred), 4))
print("NPV         :", round(m["tn"] / (m["tn"] + m["fn"]) * 100, 2), "%")
print("Brier       :", round(brier_score_loss(y, p), 4))
print("Kappa       :", round(cohen_kappa_score(y, pred), 4))
print("Confusion   : TN", m["tn"], "FP", m["fp"], "FN", m["fn"], "TP", m["tp"])
```
